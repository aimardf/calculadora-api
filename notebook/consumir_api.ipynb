# Consumir la API Calculadora desde Google Colab

Este notebook llama a la API (que corre en Docker en mi ordenador) a través del túnel público de **ngrok**, usando la librería `requests`.

**Antes de ejecutar:** arranca el contenedor y ngrok, y pega abajo en `BASE_URL` la URL que te da ngrok.

```
import requests

# Pega aquí la URL pública que muestra ngrok (sin barra al final)
BASE_URL = "https://bullpen-partner-capable.ngrok-free.dev"

# Esta cabecera evita la página de aviso de ngrok en el plan gratuito
HEADERS = {"ngrok-skip-browser-warning": "true"}
```

## 1. GET `/` — comprobar que la API responde

```
r = requests.get(f"{BASE_URL}/", headers=HEADERS)
print("Código de estado:", r.status_code)
print(r.json())
```

## 2. GET `/calcular/{operacion}` — parámetros en la URL
Calculamos `5 + 3`. La operación va en la ruta y los números como parámetros `?a=5&b=3`.

```
r = requests.get(f"{BASE_URL}/calcular/sumar", params={"a": 5, "b": 3}, headers=HEADERS)
print("URL llamada:", r.url)
print("Código de estado:", r.status_code)
print(r.json())
```

## 3. POST `/calcular` — datos en un JSON
Calculamos `8 / 2` enviando un cuerpo JSON.

```
datos = {"a": 8, "b": 2, "operacion": "dividir"}
r = requests.post(f"{BASE_URL}/calcular", json=datos, headers=HEADERS)
print("Código de estado:", r.status_code)
print(r.json())
```

## 4. Caso de error — dividir entre 0
La API responde con código **400** y un mensaje explicativo.

```
r = requests.post(f"{BASE_URL}/calcular", json={"a": 1, "b": 0, "operacion": "dividir"}, headers=HEADERS)
print("Código de estado:", r.status_code)
print(r.json())
```
